```python
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import geopandas as gpd
import icepyx as ipx
from shapely.geometry import box
import xarray as xr
```

```python
# 1. Define the bounding box bounds [X_min, Y_min, X_max, Y_max]
# xmin, ymin, xmax, ymax = [400000, -1150000, 560000, -980000]
xmin, ymin, xmax, ymax = [-25.820992, 78.232127, -19.035931, 79.607935]

# 2. Create the box geometry (automatically handles all 4 corners and closes)
polygon = box(xmin, ymin, xmax, ymax)

# 3. Add attribute data
data = {
    'id': [1],
    'name': ['ZI 79N'],
    'geometry': [polygon]
}

# 4. Create GeoDataFrame using EPSG:3413 (WGS 84 / NSIDC Sea Ice Polar Stereographic North)
gdf = gpd.GeoDataFrame(data, crs="EPSG:4326")
# gdf = gdf.set_crs('EPSG:4326', allow_override=True)

# 5. Export to Shapefile
gdf.to_file("polar_polygon.shp", driver="ESRI Shapefile")
```
