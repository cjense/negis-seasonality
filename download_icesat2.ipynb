```python
fs = s3fs.S3FileSystem(
    key=os.environ["AWS_ACCESS_KEY_ID"],
    secret=os.environ["AWS_SECRET_ACCESS_KEY"],
    client_kwargs={"endpoint_url": os.environ["S3_ENDPOINT_URL"]},
    config_kwargs={
        "request_checksum_calculation": "when_required",
        "response_checksum_validation": "when_required",
                }
)

storage_options = {
    "client_kwargs": {"endpoint_url": os.environ["S3_ENDPOINT_URL"]},
    "config_kwargs": {
        "request_checksum_calculation": "when_required",
        "response_checksum_validation": "when_required",
    },
}
```

```python
# Define the bounding box bounds [X_min, Y_min, X_max, Y_max]
# xmin, ymin, xmax, ymax = [400000, -1150000, 560000, -980000] (Polar stereographic EPSG:3413)
xmin, ymin, xmax, ymax = [-25.820992, 78.232127, -19.035931, 79.607935]

# Create the box geometry (automatically handles all 4 corners and closes)
polygon = box(xmin, ymin, xmax, ymax)

# Add attribute data
data = {
    'id': [1],
    'name': ['ZI 79N'],
    'geometry': [polygon]
}

# Create GeoDataFrame using EPSG:3413 (WGS 84 / NSIDC Sea Ice Polar Stereographic North)
gdf = gpd.GeoDataFrame(data, crs="EPSG:4326")
# gdf = gdf.set_crs('EPSG:4326', allow_override=True)

# Export to Shapefile
gdf.to_file("/tmp/polar_polygon.shp", driver="ESRI Shapefile")

# Upload to Kopah
fs.put('/tmp/polar_polygon.*', '/gaia/cjense/data/elevation/')
```
